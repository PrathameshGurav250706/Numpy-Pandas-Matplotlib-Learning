```python
# '''Handling  Missing Data'''

# df.dropna(axis =0 , inplace=True)     -->axis =0 (remove rows)  -->axis=1 (remove  columns)

import pandas as pd
data={
    "Name":['Prasad',None,"Sanket","Prathamesh","Atharav","Rahul","Sojal","Pratik"],
    "Age":[10,None,13,15,12,15,14,14],
    "Salary":[100,None,103,105,103,104,106,104],
    "Performance Score":[85,None,45,26,49,12,79,99]
}

df=pd.DataFrame(data)
print(df)

df.dropna(axis=0,inplace=True)
print(df)
```

```python
'''fillna(value , inplace=True)'''
import pandas as pd
data={
    "Name":['Prasad',None,"Sanket","Prathamesh","Atharav","Rahul","Sojal","Pratik"],
    "Age":[10,None,13,15,12,15,14,14],
    "Salary":[100,None,103,105,103,104,106,104],
    "Performance Score":[85,None,45,26,49,12,79,99]
}

df=pd.DataFrame(data)
# print(df)

# df.fillna( 0 , inplace=True)

df["Age"].fillna(df["Age"].mean() , inplace=True)
df["Salary"].fillna(df["Salary"].mean() , inplace=True)
print(df)
```

```python
import pandas as pd
data={
    "Time":[1,2,3,4,5],
    "Value":[10,None,30,40,None]
}

df=pd.DataFrame(data)
print("Before interpretation")
print(df)

print("After interpretation")
df["Value"]=df["Value"].interpolate(method="linear")
print(df)

#After interpretation
#    Time  Value
# 0     1   10.0
# 1     2   20.0     <---
# 2     3   30.0
# 3     4   40.0
# 4     5   40.0    <--
```

***Grouping**

```python
import pandas as pd
data={
    "Name":['Prasad',"Sanket","Prathamesh","Atharav","Rahul","Sojal","Pratik"],
    "Age":[10,13,10,13,15,14,14],
    "Salary":[100,103,105,108,104,106,104],
    "Performance Score":[85,45,26,49,12,79,99]
}

df=pd.DataFrame(data)

# grouped= df.groupby("Age")["Salary"].sum()
# print(grouped)

''' 
df.groupby('Age')
Age = 10 >[100,103]
Age = 13 [103,108]
Age = 15 [104]
Age = 14 [106,104]

['Salary].sum
df.groupby('Age')
Age = 10 >[100,103] =205
Age = 13 [103,108] =211
Age = 15 [104] =210
Age = 14 [106,104] =104

'''

import pandas as pd
data={
    "Name":['Prasad',"Sanket","Prathamesh","Atharav","Rahul","Sanket","Pratik"],
    "Age":[10,13,10,13,15,14,14],
    "Salary":[100,103,105,108,104,106,104],
    "Performance Score":[85,45,26,49,12,79,99]
}

df=pd.DataFrame(data)

grouped= df.groupby(['Age','Name'])["Salary"].sum()
print(grouped)
```
