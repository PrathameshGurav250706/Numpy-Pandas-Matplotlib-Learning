```python
import pandas as pd
data={
    "Name":['Prasad',"Manoj","Sanket","Prathamesh","Atharav","Rahul","Sojal","Pratik"],
    "Age":[10,12,13,15,12,15,14,14],
    "Salary":[100,102,103,105,103,104,106,104],
    "Performance Score":[85,78,45,26,49,12,79,99],
    "City":['kol','pune','mumbai','delhi','kolkata','tamilnadu','keral','jaipur']
}

df=pd.DataFrame(data)

print("Adding NaN to 'City' for the first row and filling missing values:")
df.loc[0, 'City'] = pd.NA
df.loc[3,'Salary'] =pd.NA
print(df)
```

Merge another data

```python
import pandas as pd
data={
    "Name":['Prasad',"Manoj","Sanket","Prathamesh","Atharav","Rahul","Sojal","Pratik"],
    "Age":[10,12,13,15,12,15,14,14],
    "Salary":[100,102,103,105,103,104,106,104],
    "Performance Score":[85,78,45,26,49,12,79,99],
    "City":['kol','pune','mumbai','delhi','kolkata','tamilnadu','keral','jaipur']
}

df=pd.DataFrame(data)

data2={
    "City":['kol','pune','mumbai','delhi','kolkata','tamilnadu','keral','jaipur'],
    'id':['a','b','c','d','e','f','g','h'],
    'Department': ['HR', 'IT', 'Sales', 'Finance', 'Marketing','Science','Commerce','Maths']
}
df2=pd.DataFrame(data2)

df3=pd.merge(df,df2,on="City",how='inner')
print(df3)
```
